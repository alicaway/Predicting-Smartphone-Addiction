# Predicting Smartphone Addiction

https://www.kaggle.com/competitions/playground-series-s6e8

Target: addicted_label

Metric: Area under the ROC curve (AUROC)

```python
# Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")
```

```python
# import torch
# print(torch.cuda.is_available())  # should print True
# print(torch.cuda.get_device_name(0))
```

```python
from autogluon.tabular import TabularPredictor
predictor = TabularPredictor(label="addicted_label", eval_metric="roc_auc").fit(train_df, time_limit=60)
predictions = predictor.predict_proba(test_df)

predictions_df = pd.DataFrame({'id': test_df['id'], 'addicted_label': predictions[1]})
predictions_df.to_csv('submission.csv', index=False)

predictor.leaderboard()
```
